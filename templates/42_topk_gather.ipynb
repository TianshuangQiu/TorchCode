# 🟡 Medium: Top-k Gather

Given `scores` of shape `(B, N)` and `values` of shape `(B, N, D)`, return the values corresponding to the **top-k scores per row**, ordered highest-to-lowest. Output shape: `(B, k, D)`.

This is the core pattern for retrieval, top-k routing in Mixture-of-Experts, and beam search hypothesis selection.

### Signature
```python
def topk_gather(scores: torch.Tensor, values: torch.Tensor, k: int) -> torch.Tensor:
    ...
```

### Rules
- Do **NOT** use Python `for` loops over the batch
- `1 <= k <= N`
- Output must be ordered by descending score

### Example
```
scores = [[0.1, 0.9, 0.4, 0.7]]         # (1, 4)
values = [[[1,1],[2,2],[3,3],[4,4]]]     # (1, 4, 2)
k = 2
output: [[[2,2],[4,4]]]                  # (1, 2, 2)  — top scores: 0.9 (idx=1), 0.7 (idx=3)
```

> **Reduction step (say this before coding):** `output[b, i]` is `values[b, j]` where `j` is the index of the `i`-th largest score in row `b`.
