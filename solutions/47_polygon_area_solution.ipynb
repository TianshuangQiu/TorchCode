# 🟢 Solution: Polygon Area (Shoelace)

**Primitive:** `np.roll` (cyclic vertex shift)

**Reduction:** `area = 0.5 * |Σᵢ (x[i]*y[i+1] − x[i+1]*y[i])|` — the shoelace formula, vectorised with `np.roll(..., -1)` to get the next-vertex array.

```python
# primitive: np.roll (cyclic vertex shift, shoelace formula)

import numpy as np

def polygon_area(vertices):
    vertices = np.asarray(vertices, dtype=float)
    x, y = vertices[:, 0], vertices[:, 1]
    return 0.5 * abs(np.sum(x * np.roll(y, -1) - np.roll(x, -1) * y))
```

```python
# 🔍 Verify solution
unit_square = np.array([[0.0, 0.0], [1.0, 0.0], [1.0, 1.0], [0.0, 1.0]])
print("unit square area:", polygon_area(unit_square))   # expect 1.0

triangle = np.array([[0.0, 0.0], [4.0, 0.0], [0.0, 3.0]])
print("triangle area:", polygon_area(triangle))          # expect 6.0

# same polygon in CW order
cw = unit_square[::-1].copy()
print("CW square area:", polygon_area(cw))               # expect 1.0
```

```python
# ✅ Inline test suite
import numpy as np, time

# ── Test 1: unit square ────────────────────────────────────────────────────
sq = np.array([[0.,0.],[1.,0.],[1.,1.],[0.,1.]])
assert abs(polygon_area(sq) - 1.0) < 1e-9, f"Unit square: {polygon_area(sq)}"
print("Test 1 passed: unit square")

# ── Test 2: right triangle ─────────────────────────────────────────────────
tri = np.array([[0.,0.],[4.,0.],[0.,3.]])
assert abs(polygon_area(tri) - 6.0) < 1e-9, f"Triangle: {polygon_area(tri)}"
print("Test 2 passed: right triangle (area=6)")

# ── Test 3: CW == CCW absolute area ───────────────────────────────────────
rect = np.array([[0.,0.],[3.,0.],[3.,2.],[0.,2.]])
assert abs(polygon_area(rect) - polygon_area(rect[::-1].copy())) < 1e-9, "CW vs CCW mismatch"
assert abs(polygon_area(rect) - 6.0) < 1e-9, f"Rect area: {polygon_area(rect)}"
print("Test 3 passed: CW and CCW give same absolute area")

# ── Test 4: regular hexagon ────────────────────────────────────────────────
n=6; a=np.linspace(0,2*np.pi,n,endpoint=False)
hex_pts = np.stack([np.cos(a), np.sin(a)], axis=1)
expected = 3*np.sqrt(3)/2
assert abs(polygon_area(hex_pts) - expected) < 1e-9, f"Hexagon: {polygon_area(hex_pts):.6f} vs {expected:.6f}"
print("Test 4 passed: regular hexagon")

# ── Test 5: large V=100000 ─────────────────────────────────────────────────
n=100_000; a=np.linspace(0,2*np.pi,n,endpoint=False)
circle = np.stack([np.cos(a), np.sin(a)], axis=1)
t0=time.time(); result=polygon_area(circle); elapsed=time.time()-t0
assert abs(result - np.pi) < 0.001, f"Circle approx area: {result}"
assert elapsed < 1.0, f"Too slow: {elapsed:.2f}s"
print(f"Test 5 passed: V=100000 ({elapsed:.3f}s)")

print("\nAll tests passed!")
```
